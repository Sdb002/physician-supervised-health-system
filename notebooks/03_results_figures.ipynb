# 03 — Results & Figures

Exports figures and tables straight to `thesis/figures/` and `thesis/results/`.

Requires `pip install matplotlib pandas` (analysis-only dependencies — not part
of the runtime app).

```python
import sys
from pathlib import Path

PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
sys.path.insert(0, str(PROJECT_ROOT))
FIGURES = PROJECT_ROOT / "thesis" / "figures"
RESULTS = PROJECT_ROOT / "thesis" / "results"
FIGURES.mkdir(parents=True, exist_ok=True)
RESULTS.mkdir(parents=True, exist_ok=True)
```

## Validation table → thesis/results
Exports the risk-scoring threshold table as CSV for the thesis.

```python
import csv

from tests.test_risk_scoring import (
    DIASTOLIC_CASES, GLUCOSE_CASES, HEMOGLOBIN_CASES, SYSTOLIC_CASES)
from src.clinical.risk_scoring import score_biomarker

rows = []
for value, label, risk in GLUCOSE_CASES:
    flag = score_biomarker("fasting_glucose", value)
    rows.append(["fasting_glucose", value, label, risk, flag.label == label])
for value, label, risk in SYSTOLIC_CASES:
    flag = score_biomarker("systolic_bp", value)
    rows.append(["systolic_bp", value, label, risk, flag.label == label])
for value, label, risk in DIASTOLIC_CASES:
    flag = score_biomarker("diastolic_bp", value)
    rows.append(["diastolic_bp", value, label, risk, flag.label == label])
for value, sex, pregnant, label, risk in HEMOGLOBIN_CASES:
    flag = score_biomarker("hemoglobin", value, sex, pregnant)
    rows.append([f"hemoglobin ({sex}{', pregnant' if pregnant else ''})",
                 value, label, risk, flag.label == label])

out = RESULTS / "risk_scoring_validation.csv"
with open(out, "w", newline="", encoding="utf-8") as f:
    writer = csv.writer(f)
    writer.writerow(["biomarker", "value", "expected_label", "expected_risk", "engine_agrees"])
    writer.writerows(rows)
print(f"wrote {len(rows)} validation rows to {out.relative_to(PROJECT_ROOT)}")
```
